```python
df['EngagementIndex'] = (
    df['JobInvolvement']
    + df['JobSatisfaction']
    + df['EnvironmentSatisfaction']
    + df['RelationshipSatisfaction']
) / 4
```

```python
df['EngagementIndex'].describe()
```

```python
round(df['EngagementIndex'].mean(),2)
```

## KPI 1: Engagement Index

The average Engagement Index across the organization is 2.72 out of 4.

This indicates a moderate level of employee engagement. While employees demonstrate reasonable involvement and satisfaction, there is significant room for improvement in workplace experience and employee well-being.

```python

df['BurnoutRisk'] = np.where(
    (df['OverTime'] == 'Yes') &
    (df['WorkLifeBalance'] <= 2),
    'High',
    'Low'
)
```

```python
df['BurnoutRisk'].value_counts()
```

```python
burnout_rate = (
    (df['BurnoutRisk'] == 'High').sum()
    / len(df)
) * 100

round(burnout_rate,2)
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x='BurnoutRisk',
    data=df
)

plt.title('Burnout Risk Distribution')
plt.show()
```

## Burnout Risk Analysis

### Key Findings

| Metric | Value |
|----------|----------|
| Total Employees | 1470 |
| High Burnout Risk Employees | 126 |
| Low Burnout Risk Employees | 1344 |
| Burnout Risk Rate | 8.57% |

### Interpretation

- 8.57% employees are classified as High Burnout Risk.
- These employees work overtime and report low work-life balance.
- This segment is more vulnerable to disengagement and future attrition.
- HR should prioritize workload management and well-being initiatives for this group.
