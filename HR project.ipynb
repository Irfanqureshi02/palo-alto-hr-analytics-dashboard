```python

pd.crosstab(
    df['BurnoutRisk'],
    df['Attrition']
)
```

```python
burnout_attrition = pd.crosstab(df['BurnoutRisk'], df['Attrition'])

burnout_attrition_pct = burnout_attrition.div(
    burnout_attrition.sum(axis=1),
    axis=0
) * 100

round(burnout_attrition_pct,2)
```

Employees identified as High Burnout Risk have an attrition rate of 33.33%,
which is more than double the attrition rate of Low Burnout Risk employees (14.51%).

This suggests a strong relationship between burnout and employee exits.
Employees working overtime while experiencing poor work-life balance are significantly
more likely to leave the organization.

The finding supports the project's core hypothesis that burnout often precedes attrition
and should be monitored as an early-warning indicator for proactive HR intervention.

The analysis identified several early-warning indicators of employee disengagement and burnout.
